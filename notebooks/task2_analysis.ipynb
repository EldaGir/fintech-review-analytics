```python
# Task 2 — Sentiment and Thematic Analysis

# This notebook documents the sentiment analysis and thematic analysis pipeline used to analyze Ethiopian banking app reviews collected from the Google Play Store.
```

```python
# ## Sentiment Analysis Methodology
#
# Two sentiment analysis approaches were implemented:
#
# 1. VADER Sentiment Analysis
#    - Lexicon-based sentiment analysis
#    - Produces positive, neutral, and negative sentiment scores
#
# 2. DistilBERT Transformer Model
#    - Model: `distilbert-base-uncased-finetuned-sst-2-english`
#    - Pretrained transformer-based sentiment classifier
#    - Produces sentiment labels and confidence scores
#
# DistilBERT was selected as the primary sentiment model because transformer models generally outperform traditional lexicon-based approaches in contextual understanding and sentiment classification accuracy.
```

```python
# ## NLP Preprocessing Pipeline
#
# The preprocessing pipeline included:
#
# - Lowercase normalization
# - Removal of special characters
# - Tokenization
# - Stop-word removal
# - Text cleaning using regular expressions
#
# The cleaned text was used for both sentiment analysis and thematic analysis.
```

```python
# ## Theme Identification and Grouping Logic
#
# Themes were identified using TF-IDF keyword extraction and rule-based keyword grouping.
#
# Recurring keywords and bigrams were analyzed to identify major business-related issues discussed by users.
#
# The following themes were defined:
#
# | Theme | Description |
# |---|---|
# | Account Access Issues | Login failures, OTP problems, verification issues |
# | Transaction Performance | Slow transfers, payment delays, transaction failures |
# | App Stability & Bugs | Crashes, errors, bugs, broken functionality |
# | UI & Design | Navigation, interface usability, user experience |
# | Customer Support | Service responsiveness and customer assistance |
# | Feature Requests | Requests for additional features and updates |
#
# Reviews that did not clearly match a predefined business issue category were classified as "Other".
```

```python
# ## Key Findings
#
# ### Sentiment Analysis
#
# - Most reviews were classified as positive.
# - DistilBERT successfully assigned sentiment labels and confidence scores to more than 90% of reviews.
# - Positive reviews were dominated by words such as "good", "best", and "excellent".
#
# ### Thematic Analysis
#
# The most common operational issues identified were:
#
# - Transaction performance problems
# - Application stability and bug-related complaints
# - Account access and verification issues
#
# Many reviews contained short generic praise messages that did not map to actionable business themes and were therefore categorized as "Other".
```
